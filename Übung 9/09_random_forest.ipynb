# Random Forest

In dieser Übung wird ein Datensatz von [kaggle.com](https://www.kaggle.com/datasets/mathchi/diabetes-data-set?select=diabetes.csv) verwendet, um das Diabetesrisiko vorherzusagen.

Lesen Sie den Datensatz ein und schauen Sie ihn sich an. Praktischerweise sind alle Daten bereits numerisch. Da Random Forests nicht anfällig für schlecht skalierte Daten sind, kann man auf Feature Engineering verzichten.

````python
import pandas as pd

data = pd.read_csv("diabetes.csv")
data.describe()
````

Trainieren Sie nun einen ```RandomForestClassifier``` auf Basis aller Eingangsdaten, um Diabetes vorherzusagen. Diese Information steht in der Spalte "Outcome". Alle anderen Spalten kann man direkt als Feature verwenden. Achten Sie darauf, die Spalte "Outcome" nicht als Feature zu nutzen! Betrachten Sie anschließend die Confusion Matrix und berechnen Sie die Accuracy.

# TDOO:
## описать идею этого классификатора по памяти

````python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, accuracy_score

feature_cols = ["Pregnancies", "Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI", "DiabetesPedigreeFunction", "Age"]

### ADD CODE HERE
features = data[feature_cols]
labels = data[["Outcome"]]

def train_model(X, Y):
    model = RandomForestClassifier()
    model.fit(X, Y)
    return model

model = train_model(features, labels)


def get_metrics(model, features, labels):
    preds = model.predict(X=features)
    acc = accuracy_score(labels, preds)
    confusion_mat = confusion_matrix(y_true=labels, y_pred=preds)
    print(f"Metrics:\n+ Acc: {acc}\n+ Confusion matrix:\n {confusion_mat}\n")



get_metrics(model, features, labels)
````

Bei Training mit allen Daten erreicht ein ```RandomForestClassifier``` eine Accuracy von 100%. Auch bei Machine Learning sind Dinge, die zu gut klingen, um wahr zu sein, meist auch nicht wahr. Die Vermutung, dass man overfitted, liegt nahe.

Mit der Funktion ```train_test_split``` kann man die Daten in mehrere Teil eteilen. Mit dem Parameter ```test_size=``` kann man angeben, welcher Anteil der Daten als Test-Datensatz verwendet werden soll. 

Teilen Sie die Daten in 80% Trainings- und 20% Testdaten auf. Trainieren Sie den Random Forest anschließend erneut, diesmal aber nur mit den Trainingsdaten. Berechnen Sie Accuracy und Confusion Matrix auf Basis sowohl der Test-, als auch der Trainingsdaten.

````python
from sklearn.model_selection import train_test_split

### ADD CODE HERE

features_train, features_test, labels_train, labels_test = train_test_split(features, labels, test_size=0.2, shuffle=True, random_state=1)

model = train_model(features_train, labels_train)

print("train")
get_metrics(model, features_train, labels_train)
print("test")
get_metrics(model, features_test, labels_test)
````
